```python
import geopandas as gpd
```

### Calculated the suitability of good land per hectare for each city.

```python
aptitude = gpd.read_file("./soil/Classe_aptidao_agricola.shp")
municipies_shp = gpd.read_file("./soil/MT_Municipios_2025.shp")

municipies_shp = municipies_shp.to_crs(aptitude.crs)

municipies_shp['mun_total_area'] = municipies_shp.geometry.area

intersection = gpd.overlay(municipies_shp, aptitude, how='intersection')

intersection['intersect_area'] = intersection.geometry.area

intersection['area_ha'] = intersection.geometry.area / 10000

posology = intersection.pivot_table(
    index=['CD_MUN', 'NM_MUN'],  
    columns='descricao_',         
    values='area_ha', 
    aggfunc='sum'
).fillna(0).round(2).reset_index()                    

df_posology_final = posology[['CD_MUN', 'Terra Com Aptidao Boa', 'NM_MUN']].rename(columns={'CD_MUN': 'code', 'NM_MUN': 'city', "Terra Com Aptidao Boa": "good_aptitude_ha"})

df_posology_final['code'] = df_posology_final['code'].astype(int)
```
